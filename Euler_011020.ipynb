```python
# This is Euler project's problem 17
first_numbers = ['','one','two','three','four','five','six','seven','eight','nine','ten',
                 'eleven','twelve','thirteen','fourteen','fifteen','sixteen','seventeen','eighteen','nineteen']
tens = ['','','twenty','thirty','forty','fifty','sixty','seventy','eighty','ninety']
other = ['hundred','hundredand','onethousand']
f_n_length = [len(x) for x in first_numbers]
tens_length = [len(x) for x in tens]
other_length = [len(x) for x in other]

def letters_number(n):
    n_dig = to_digits(n)
    if n < 20:
        return f_n_length[n]
    elif n < 100:
        return tens_length[n_dig[-2]] + letters_number(n % 10)
    elif n < 1000:
        if n % 100 == 0:
            return f_n_length[n_dig[-3]] + other_length[0]
        else:
            return f_n_length[n_dig[-3]] + other_length[1] + letters_number(n % 100)
    else:
        return other_length[2]

total_letter_number = 0
for i in range(1001):
    total_letter_number += letters_number(i)
total_letter_number
```

```python
# This is Euler project's problem 19

days_in_month = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
days_in_month_leap = [31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

total_sundays = 0

weekday = 1
day = 1
month = 1
year = 1900

def is_leap_year(y):
    if y % 4 == 0 and y % 100 != 0:
        return True
    elif y % 400 == 0:
        return True
    else:
        return False

while year < 2001:
    if year > 1900 and weekday == 7 and day == 1:
        total_sundays +=1
    weekday += 1
    day += 1
    if weekday > 7:
        weekday = 1
    if is_leap_year(year):
        if day > days_in_month_leap[month - 1]:
            day = 1
            month += 1
    else:
        if day > days_in_month[month - 1]:
            day = 1
            month += 1   
    if month > 12:
        month = 1
        year += 1
total_sundays
```
