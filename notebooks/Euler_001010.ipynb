```python
# This is Euler project's problem 4 
def is_palyndrome(n):
    my_digits = to_digits(n)
    j = 1
    while j <= len(my_digits)/2:
        if my_digits[-j]!= my_digits[j-1]:
            return False
        j +=1
    return True

all_products=[]
for i in range(100,1000):
    for j in range(i,1000):
        all_products.append(i*j)
all_products.sort()
all_products.reverse()

for j in all_products:
    if is_palyndrome(j):
        break
j
```

```python
# This is Euler project's problem 5
my_list = [i for i in range(1,21)]
all_divs = []
while len(my_list) > 0:
    if my_list[0]==1:
        my_list.pop(0)
    else: 
        div = find_first_prime_factor(my_list[0])
        all_divs.append(div)
        for j in range(len(my_list)):
            if my_list[j] % div == 0:
                my_list[j] = my_list[j] // div
total_product_list(all_divs)
```

```python
# This is Euler project's problem 9
def is_pythagorean(triplet):
    if triplet[0]**2 == triplet[1]**2 + triplet[2]**2:
        return True
    else:
        return False

triplets = []
for i in range(666):
    for j in range(1,i//2+1):
        my_triplet = [1000 - i, i - j, j]
        triplets.append(my_triplet)
for i in triplets:
    if is_pythagorean(i):
        print(i[0]*i[1]*i[2])
```
