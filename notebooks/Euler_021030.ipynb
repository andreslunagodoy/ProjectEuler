```python
# This is Euler project's problem 21

def euler21_function_d(n):
    return total_sum_list(proper_divisors(n))

def is_amicable(n):
    if n == euler21_function_d(euler21_function_d(n)) and n != euler21_function_d(n):
        return True
    else:
        return False
    
all_amicables=[]
for j in range(2,10000):
    if is_amicable(j):
        all_amicables.append(j)
total_sum_list(all_amicables)
```

```python
# This is Euler project's problem 22
with open("files/0022_names.txt") as names_file:
    names2 = names_file.readline()
names1 = names2.split(',')
names =[k.strip('"') for k in names1]
names.sort()

letters = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'

total_score = 0
for j in range(len(names)):
    name_score = 0
    for k in names[j]:
        name_score += letters.find(k) + 1
    total_score += (j+1)*name_score
total_score
```

```python
# This is Euler project's problem 23
def is_abundant(n):
    if total_sum_list(proper_divisors(n))>n:
        return True
    else:
        return False

all_abundants=[]
for j in range (12,28123):
    if is_abundant(j):
        all_abundants.append(j)
        
sums_of_abundants=[]
for i in range(len(all_abundants)):
    for j in range(i,len(all_abundants)):
        my_sum = all_abundants[i] + all_abundants[j]
        if my_sum < 28124:
            sums_of_abundants.append(my_sum)
        else:
            break
sums_of_abundants=list(set(sums_of_abundants))
sums_of_abundants.sort
sum_of_unbreakables=0
for i in range(1,28124):
    if i not in sums_of_abundants:
        sum_of_unbreakables += i
sum_of_unbreakables
```

```python
# This is Euler project's problem 26
with open("files/0026_digits.txt") as digits_file:
    digits2 = digits_file.readline()
digits1 = digits2.split(' ')
len(digits1)
my_precision = 1500

digits_mat = list(list(digits1[my_precision*j+i] for i in range(my_precision)) for j in range(1000))

def find_period(n,my_dig,my_len,my_dep,precision):
    period = 1
    while True:
        if my_dig[n-1][my_dep:my_dep+my_len] == my_dig[n-1][my_dep + period:my_dep+my_len + period]:
            return period
        period += 1
        if period > precision:
            return period
        
long_guys = []
max_period = 0
seed_for_period = 0
for x in range(1,1001):
    period = find_period(x,digits_mat,10,100,my_precision)
    if period > max_period:
        max_period = period
        seed_for_period = x
    if period > my_precision:
        long_guys.append(x)
print(long_guys)
print([seed_for_period,max_period])
```

```python
# This is Euler project's problem 27    
candidates=[]
for b in primes_under(1000):
    for j in primes_under(2000):
        if j-b>1000:
            break
        else:
            candidates.append([j-b-1,b])

def prime_sequence_length(cand):
    j=0
    while True:
        if is_prime_aux(j*j+cand[0]*j+cand[1]):
            j=j+1
        else:
            return j
        
longest_sequence=0;
for i in range(len(candidates)):
    seq = prime_sequence_length(candidates[i])
    if seq > longest_sequence:
        longest_sequence=seq
        longest_candidate=candidates[i]
longest_candidate[0]*longest_candidate[1]
```
