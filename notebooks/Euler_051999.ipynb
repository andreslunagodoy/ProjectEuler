```python
# This is Euler project's problem 51
index = 4
finish = False
while not finish:
    n = my_list_of_primes[index]
    digits_n = to_digits(n)
    max_counter = 0
    for rep_set in all_subsets(range(len(digits_n) - 1)):
        if len(set([digits_n[k] for k in rep_set])) == 1:
            new_numbers = []
            for digit in range(10):
                if digit == 0 and 0 in rep_set:
                    continue
                new_num_digits = digits_n.copy()
                for position in rep_set:
                    new_num_digits[position] = digit
                new_num = digits_to_integer(new_num_digits)
                new_numbers.append(new_num)
            counter = 0
            for num in new_numbers:
                if is_prime_aux(num):
                    counter += 1
            if counter > max_counter:
                max_counter = counter
                max_rep = rep_set
        if max_counter > 7:
            print([n,max_counter,max_rep])
            finish = True
            break
    index +=1
```

```python
# This is Euler project's problem 56
def reverse_number(n):
    digits = to_digits(n)
    digits_rev = digits.copy()
    digits_rev.reverse()
    return digits_to_integer(digits_rev)

def is_lychrel(n):
    num = n
    for k in range(50):
        num += reverse_number(num)
        if is_palyndrome(num):
            return False
    return True

total_lychrel = 0
for j in range(1,10000):
    if is_lychrel(j):
        total_lychrel += 1
total_lychrel
```
